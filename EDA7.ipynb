```python
import numpy as np
import pandas as pd
```

```python
index=[('a',2010),('a',2020),('a',2025),('b',2010),('b',2020),('b',2025)]
population=[123,456,789,101,121,131]
i=pd.MultiIndex.from_tuples(index)
print(i)
pop=pd.Series(population,index=i)
print(i)
pop=pop.reindex(index=i)
print(pop)
pop_df=pop.unstack()
print(pop_df)
```

```python
index=[('a',2010),('a',2020),('a',2025),('b',2010),('b',2020),('b',2025)]
population=[123,456,789,101,121,131]
i1=pd.MultiIndex.from_arrays([['a','a','a','b','b','b'],[2010,2020,2025,2010,2020,2025]],names=["country","year"])
pop1=pd.Series(population,index=i1)
print(pop1)
```

```python
i2=pd.MultiIndex.from_product([['a','b'],[2010,2020,2025]],names=["country","year"])
pop2=pd.Series(population,index=i2)
print(pop2)
```

```python
i3=pd.MultiIndex.from_arrays([['A','B'],[2010,2020]],names=["country","year"])
pop3=pd.Series([123,456],index=i3)
print(pop3)
```
