```python
import numpy as np 
print(np.random.seed(1))
arr=np.random.randint(10,size=(2,5))
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
print(arr.itemsize)
print(arr.nbytes)
print(arr[1,1])
print(arr[-1,-2])
print(arr[:2,:2])
print(arr[:,0])
print(arr[:0,])
```

```python
np.random.seed(1)
arr=np.random.randint(10,size=10)
print(arr)
print(arr[3])
print(arr[-2])
print(arr[3:9:2])
print(arr[::-1])
```
